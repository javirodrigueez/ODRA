```python
import json
import os
```

```python
folder_path = '../eval_magnet'
```

```python
ctrs_list = ["general_1_understanding", "general_2_interpersonal_effectiveness", "general_3_collaboration", "CBT_1_guided_discovery", "CBT_2_focus", "CBT_3_strategy"]
```

## Using one folder

```python
for j in ctrs_list:
    total_rating = 0
    count = 0
    for filename in os.listdir(folder_path):
        if filename.endswith('.json'):
            file_path = os.path.join(folder_path, filename)
            with open(file_path, 'r', encoding='utf-8') as f:
                json_data = json.load(f)
            total_rating = total_rating + json_data[j]
            count = count + 1
    avg_rating = total_rating/count
    print(j + ":")
    print(avg_rating)
```

## Using various folders

```python
folder_path_1 = '../eval_ds_attitudes/ctrs/neutral'
folder_path_2 = '../eval_ds_attitudes/ctrs/positive'

for j in ctrs_list:
    total_rating = 0
    count = 0
    for filename in os.listdir(folder_path_1):
        if filename.endswith('.json'):
            file_path = os.path.join(folder_path_1, filename)
            with open(file_path, 'r', encoding='utf-8') as f:
                json_data = json.load(f)
            total_rating = total_rating + json_data[j]
            count = count + 1
    for filename in os.listdir(folder_path_2):
        if filename.endswith('.json'):
            file_path = os.path.join(folder_path_2, filename)
            with open(file_path, 'r', encoding='utf-8') as f:
                json_data = json.load(f)
            total_rating = total_rating + json_data[j]
            count = count + 1
    avg_rating = total_rating/count
    print(j + ":")
    print(avg_rating)
```

## Excluding from conversation file (from one folder and its reference)

```python
folder_path = '../eval_magnet'
other_folder_path = '../../out/magnet_sessions'
for j in ctrs_list:
    total_rating = 0
    count = 0
    
    for filename in os.listdir(folder_path):
        if filename.endswith('.json'):
            # 1. Define the path for the file in the other folder
            other_file_path = os.path.join(other_folder_path, filename)
            
            # 2. Check if the matching file actually exists before trying to open it
            if not os.path.exists(other_file_path):
                continue
                
            # 3. Open the file in the other folder and check the attitude
            with open(other_file_path, 'r', encoding='utf-8') as f_other:
                other_data = json.load(f_other)
                
            # Use .get() to safely check the key, avoiding KeyError if 'attitude' is missing
            if other_data.get('AI_client').get('attitude') == 'negative':
                
                # 4. If positive, proceed with your original logic
                file_path = os.path.join(folder_path, filename)
                with open(file_path, 'r', encoding='utf-8') as f_main:
                    json_data = json.load(f_main)
                    
                total_rating += json_data[j]
                count += 1

# Safeguard against division by zero
    if count > 0:
        avg_rating = total_rating / count
        print(f"{j}:\n{avg_rating}")
    else:
        print(f"{j}:\nNo positive attitude files found.")
```
