```python
import json
import os
```

```python
folder_path = 'out_test'
```

```python
avg_wais = []
for j in range(12):
    total_rating = 0
    count = 0
    for filename in os.listdir(folder_path):
        if filename.endswith('.json'):
            file_path = os.path.join(folder_path, filename)
            with open(file_path, 'r', encoding='utf-8') as f:
                json_data = json.load(f)
            total_rating = total_rating + json_data["wai"+str(j+1)]
            count = count + 1
    avg_rating = total_rating/count
    avg_wais.append(avg_rating)
```

```python
score_task = (avg_wais[0] + avg_wais[1] + avg_wais[7] + avg_wais[11])/4
score_goal = ((8 - avg_wais[3]) + avg_wais[5] + (8 - avg_wais[9]) + avg_wais[10])/4
score_bond = (avg_wais[2] + avg_wais[4] + avg_wais[6] + avg_wais[8])/4

print("score_task: " + str(score_task))
print("score_goal: " + str(score_goal))
print("score_bond: " + str(score_bond))
```
